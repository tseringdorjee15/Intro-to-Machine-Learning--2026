```python
import statsmodels.api as sm    # Loads tools for running statistical tests and regressions
from sklearn import datasets    # Brings in scikit-learn's built-in data utilities
import pandas as pd             # Imports the standard data manipulation library

# RAW github links downloaded and stored here in 2 variables
train_url = 'https://raw.githubusercontent.com/tseringdorjee15/Intro-to-Machine-Learning--2026/refs/heads/main/Boston-Housing/train.csv'
test_url = 'https://raw.githubusercontent.com/tseringdorjee15/Intro-to-Machine-Learning--2026/refs/heads/main/Boston-Housing/test.csv'

# datasets loaded directly into DataFrames
df_train = pd.read_csv(train_url)
df_test = pd.read_csv(test_url)

# Preview data to verify if it is loaded correctly or not
print("--- Training Data Preview ---")
print(df_train.head())                       # print the first 5 rows
print("--- Testing Data Preview ---")
print(df_test.head())                        # print the first 5 rows
```

```python
# Split datasets into two separate pieces so that a machine learning model can learn from them.

import numpy as np
import pandas as pd

# Set the features (create a new dataframe named "X_train". It takes original dataset (df_train) and copies everything except the 'medv' (housing price) column.)
X_train = df_train.drop(columns=['medv'])

# Set the target, new dataframe named "Y_train". It isolates just the single column 'medv' into its own separate list
y_train = df_train['medv']
```

```python
# SINGLE LINEAR REGRESSION IN "Statsmodels"
# Builds, trains, and tests a Simple Linear Regression model using the statsmodels library to look for a statistical relationship between two variables.

import statsmodels.api as sm

# 1. Using correct DataFrame (df_train) and column name
X = df_train["rm"]           # Sets the number of rooms (rm) as a single predictor feature

y = df_train["medv"]         # Sets the median home value (medv) as the target outcome we want to predict.

# TRAINING THE MODEL - Fit and make the predictions by the model
# sm.OLS: Stands for Ordinary Least Squares. This is the formal math name for a classic linear regression. It calculates a straight "line of best fit"
# through data points..fit(): Tells Python to run the math equations over your training data right now to calculate the weight/slope of the line.
model = sm.OLS(y, X).fit()

# Passes the room counts (X) back into a trained line equation to get the model's generated price guesses.
predictions = model.predict(X)

# Print out the statistics - This is the most important command. It outputs a large, highly detailed statistical report showing our model's accuracy
# (like \(R^{2}\)), coefficients (slopes), and \(p\)-values.
model.summary()
```

```python
X = sm.add_constant(X)     # Addition of the constant term has a direct effect on the coefficient term.  Without it, a model was passing through the origin,
                           # but now we have a y-intercept at -34.67. Also the slope of the RM predictor is also changed from 3.634 to 9.1021 (coef of RM).

model = sm.OLS(y, X).fit()
predictions = model.predict(X)

model.summary()
```

```python
# MULTIPLE LINEAR REGRESSION IN "Statsmodels"

# Now we will fit a regression model with more than one variable — "lstat" (percentage of lower status of the population) variable along with the "rm" variable.
# The model training (fitting) procedure remains the exact same as previous:

X = df_train[["rm", "lstat"]]
y = df_train["medv"]

model = sm.OLS(y, X).fit()
predictions = model.predict(X)

model.summary()
```

**RUNNING 2 LINEAR REGRESSION MODELS SIDE BY SIDE:**

```python
# MODEL 1- (ALL 13 [FEATURES] EXCLUDING 'medv'[target]) - Trains a multiple linear regression model to predict housing prices.

import statsmodels.api as sm

# Select ALL columns except the target 'medv'
X_all = df_train.drop(columns=['medv'])
X_all = sm.add_constant(X_all) # Add intercept
y = df_train['medv']

model_all = sm.OLS(y, X_all).fit()
print(model_all.summary())
```

```python
# Cell for Model 2 (Subset: Rooms & Status Only) - trains a simplified regression model using a subset of only two features (rm and lstat) for comparison.

# Select only 'rm' and 'lstat' columns
X_sub = df_train[['rm', 'lstat']]
X_sub = sm.add_constant(X_sub) # Add intercept

model_sub = sm.OLS(y, X_sub).fit()
print(model_sub.summary())
```

***Model Comparison and Contrast Summary:***

**Model 1 (All 13 Features):**
    - It achieves a higher R-squared score of 0.736.
    - it captures more complex dynamics and explains an additional 10.2% of the variance in housing prices compared to the simplified model.
    
**Model 2 (Subset Features)**
    - It drops 11 variables and focuses strictly on only 2 features which are: room count (rm) and lower-class status (lstat).
    - While its R-squared score drops to 0.634, it remains a highly valuable   model because it is simple, easy to interpret, and highlights the two single most impactful drivers of real estate value without introducing noise or unnecessary complexity.
